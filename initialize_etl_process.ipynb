```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, avg, max, min
```

```python
spark = SparkSession.builder \
    .appName("ETL Model") \
    .getOrCreate()
```

```python
# Define the path to the CSV file
csv_file_path = "/content/healthcare_dataset.csv"

# Extract: Read the CSV file into a DataFrame
df = spark.read.csv(csv_file_path, header=True, inferSchema=True)
```

```python
df_filtered = df.dropna()
```

```python
# Replace 'existing_column' with the actual column name you want to use.
# For example, if you want to use the 'Billing Amount' column:
df_transformed = df_filtered.withColumn("new_column", col("Billing Amount") * 2)

# If you are unsure of the exact column names, you can print them out using:
print(df_filtered.columns)
```

```python
df_transformed.show()
```

```python
# Replace 'group_column' with an actual column from your DataFrame you want to group by.
# Replace 'numeric_column' with the column you want to perform aggregation functions on.
# For example, if you want to group by 'Hospital' and sum 'Billing Amount':
sum_df = df_transformed.groupBy("Hospital").agg(sum("Billing Amount").alias("sum_column"))
sum_df.show()

# Similarly for other aggregation functions:
avg_df = df_transformed.groupBy("Hospital").agg(avg("Billing Amount").alias("avg_column"))
avg_df.show()

max_df = df_transformed.groupBy("Hospital").agg(max("Billing Amount").alias("max_column"))
max_df.show()

min_df = df_transformed.groupBy("Hospital").agg(min("Billing Amount").alias("min_column"))
min_df.show()
```
